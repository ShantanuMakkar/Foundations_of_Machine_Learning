```python
# ==========================================
# 1. IMPORTS & DATA LOADING
# ==========================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, ConfusionMatrixDisplay

# Load the breast cancer dataset
bc_data = load_breast_cancer()

# Extract predictors and target
# Target: 0 = malignant, 1 = benign
X = bc_data.data
y = bc_data.target
feature_names = bc_data.feature_names

# ==========================================
# 2. DATA SPLITTING
# ==========================================
# 70:30 split, random_state=9001, stratified by target labels
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.30,       # 30% testing, 70% training
    random_state=9001,    # Fixed random state per instructions
    stratify=y            # Ensures equal distribution of classes in train/test
)

# ==========================================
# 3. MODEL BUILDING & TRAINING
# ==========================================
# Initialize decision tree with max_depth=3 and random_state=9001
dt_classifier = DecisionTreeClassifier(max_depth=3, random_state=9001)

# Fit the tree on the training set
dt_classifier.fit(X_train, y_train)

# ==========================================
# 4. PREDICTIONS
# ==========================================
# Make predictions on the testing set
y_test_pred = dt_classifier.predict(X_test)
```

```python
# ---------------------------------------------------------
# Q1: Accuracy Score on the test split
# ---------------------------------------------------------
test_accuracy = accuracy_score(y_test, y_test_pred)
print(f"Test Set Accuracy: {test_accuracy * 100:.2f}%")

# ---------------------------------------------------------
# Q2: Cases of malignant tumors wrongly predicted as benign
# ---------------------------------------------------------
# In this dataset, Malignant = 0, Benign = 1
# We are looking for False Positives for the benign class (True=0, Pred=1)
cm = confusion_matrix(y_test, y_test_pred)
print("\nConfusion Matrix:")
print(cm)

# Using ConfusionMatrixDisplay for better visualization
fig, ax = plt.subplots(figsize=(5, 4))
ConfusionMatrixDisplay.from_estimator(dt_classifier, X_test, y_test, cmap=plt.cm.Blues, ax=ax)
plt.title("Test Data Confusion Matrix")
plt.show()

malignant_as_benign = cm[0, 1]
print(f"-> Malignant cases wrongly predicted as benign: {malignant_as_benign}")

# ---------------------------------------------------------
# Q3: Recall for malignant class (Class 0)
# ---------------------------------------------------------
print("\nClassification Report:")
report = classification_report(y_test, y_test_pred, digits=2)
print(report)
# You can see the recall specifically for class '0' in the printout

# ---------------------------------------------------------
# Q4: First node split (Root Node) Visualization
# ---------------------------------------------------------
print("\nVisualizing the Decision Tree...")
fig = plt.figure(figsize=(15, 8))
_ = tree.plot_tree(dt_classifier, 
                   feature_names=feature_names,
                   class_names=['Malignant', 'Benign'],
                   filled=True, 
                   rounded=True)
plt.show()

# Programmatically print the exact root node split
root_feature = feature_names[dt_classifier.tree_.feature[0]]
root_threshold = dt_classifier.tree_.threshold[0]
print(f"-> The very first split is: {root_feature} <= {root_threshold:.2f}")
```
