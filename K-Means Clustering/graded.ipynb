```python
# ==========================================
# 1. IMPORTS
# ==========================================

import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler

from sklearn.cluster import KMeans

from sklearn.metrics import silhouette_score
```

```python
# ==========================================
# 2. LOAD DATASET
# ==========================================

df = pd.read_csv("carprices-truncated.csv")

print("Dataset Shape:")

print(df.shape)

print("\nColumns:")

print(df.columns)

print("\nFirst 5 Rows:")

print(df.head())
```

```python
# ==========================================
# 3. FEATURE MATRIX
# ==========================================

X = df

print(X.shape)
```

```python
# ==========================================
# 4. FEATURE SCALING
# ==========================================

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled.shape)
```

```python
# ==========================================
# 5. K-MEANS CLUSTERING
# ==========================================

k_values = [2,3,4,5,6]

inertia_list = []
silhouette_list = []
cluster_counts = {}

for k in k_values:

    kmeans = KMeans(
        n_clusters=k,
        init='random',
        random_state=9001,
        n_init=20
    )

    kmeans.fit(X_scaled)

    labels = kmeans.labels_

    # Inertia
    inertia_list.append(kmeans.inertia_)

    # Silhouette Score
    silhouette_list.append(silhouette_score(X_scaled,labels))

    # Cluster Counts
    counts = pd.Series(labels).value_counts().sort_index()

    cluster_counts[k] = counts

    print("\n===================")
    print(f"K = {k}")
    print("===================")
    print(
        "Inertia :",
        round(kmeans.inertia_,2)
    )

    print(
        "Silhouette Score :",
        round(silhouette_score(X_scaled,labels),4)
    )

    print("\nCluster Counts:")
    print(counts)
```

```python
# ==========================================
# 6. RESULTS SUMMARY
# ==========================================

results_df = pd.DataFrame({

    'K' : k_values,

    'Inertia' : inertia_list,

    'Silhouette Score' : silhouette_list

})

results_df
```

```python
# ==========================================
# 7. ELBOW METHOD
# ==========================================

results_df['Inertia Drop'] = (
    results_df['Inertia'] - results_df['Inertia'].shift(-1)
)

print(results_df)
```

```python
# ==========================================
# 8. OPTIMAL K
# ==========================================

largest_drop = results_df['Inertia Drop'].max()

optimal_k = results_df.loc[
    results_df['Inertia Drop'].idxmax(),'K'
]

print("Largest Inertia Drop :",round(largest_drop,2))
print("\nOptimal K (Elbow Method) :",optimal_k)
```

```python
# ==========================================
# 9. ELBOW CURVE
# ==========================================

plt.figure(
    figsize=(7,5)
)

plt.plot(
    k_values,
    inertia_list,
    marker='o'
)

plt.xlabel("K")
plt.ylabel("Inertia (WCSS)")
plt.title("Elbow Method")
plt.grid()
plt.show()
```

```python
# ==========================================
# 10. SILHOUETTE SCORES
# ==========================================

plt.figure(
    figsize=(7,5)
)

plt.plot(
    k_values,
    silhouette_list,
    marker='o'
)

plt.xlabel("K")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Scores")
plt.grid()
plt.show()
```
