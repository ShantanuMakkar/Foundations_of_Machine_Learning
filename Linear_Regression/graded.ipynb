```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

```python
df = pd.read_excel('dataset.xlsx')
```

```python
# Remove the column containing the name of each university (the first column)
df = df.iloc[:, 1:]
```

```python
# Obtain descriptive statistics rounded to 2 decimal places
desc_stats = df.describe().round(2)
print(desc_stats)
```

```python
# One-hot encode the 'Private' variable, dropping the first redundant column
df_encoded = pd.get_dummies(df, columns=['Private'], drop_first=True)
```

```python
# Consider 'Grad.Rate' as the target variable
X = df_encoded.drop(columns=['Grad.Rate'])
y = df_encoded['Grad.Rate']
```

```python
# Train/Test Split (70:30) with random_state = 42
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
# Fit the model on the training set
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Output model coefficients, rounded to 2 decimal places
coef_df = pd.DataFrame({'Feature': X.columns, 'Coefficient': model.coef_}).round(2)
print(coef_df)

# Output intercept, rounded to 2 decimal places
intercept = round(model.intercept_, 2)
print(f'Intercept: {intercept}')
```

```python
# Obtain predictions on training and testing sets
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)
```

```python
# Generate metrics rounded to 2 decimal places
print("Training data ..")
print(f"Train R2: {round(r2_score(y_train, y_train_pred), 2)}")
print(f"Train MSE: {round(mean_squared_error(y_train, y_train_pred), 2)}")
print(f"Train MAE: {round(mean_absolute_error(y_train, y_train_pred), 2)}")

print("Testing data ..")
print(f"Test R2: {round(r2_score(y_test, y_test_pred), 2)}")
print(f"Test MSE: {round(mean_squared_error(y_test, y_test_pred), 2)}")
print(f"Test MAE: {round(mean_absolute_error(y_test, y_test_pred), 2)}")
```
