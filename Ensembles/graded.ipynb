```python
# ==========================================
# 1. IMPORTS
# ==========================================

import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

from sklearn.ensemble import RandomForestClassifier

from lightgbm import LGBMClassifier

from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report,
    roc_auc_score
)
```

```python
# ==========================================
# 2. LOAD DATASET
# ==========================================

bc_data = load_breast_cancer()

# Predictors
X = bc_data.data

# Target
# 0 = malignant
# 1 = benign
y = bc_data.target

feature_names = bc_data.feature_names

print("Shape of X :", X.shape)
print("Shape of y :", y.shape)

print("\nTarget Names:")
print(bc_data.target_names)
```

```python
# ==========================================
# 3. TRAIN TEST SPLIT
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=9001,
    stratify=y
)

print("Training Shape :", X_train.shape)
print("Testing Shape :", X_test.shape)
```

```python
# ==========================================
# 4. RANDOM FOREST MODEL
# ==========================================

rf_model = RandomForestClassifier(
    class_weight='balanced',
    random_state=9001,
    n_estimators=100,
    max_depth=4
)

rf_model.fit(X_train, y_train)

y_rf_pred = rf_model.predict(X_test)

# Probability predictions
y_rf_prob = rf_model.predict_proba(X_test)[:,1]
```

```python
# ==========================================
# 5. RANDOM FOREST EVALUATION
# ==========================================

rf_cm = confusion_matrix(y_test, y_rf_pred)

print("Random Forest Confusion Matrix")
print(rf_cm)


disp = ConfusionMatrixDisplay(
    confusion_matrix=rf_cm,
    display_labels=bc_data.target_names
)

disp.plot(cmap=plt.cm.Blues)
plt.title("Random Forest Confusion Matrix")
plt.show()


print("\nClassification Report\n")

print(
    classification_report(
        y_test,
        y_rf_pred,
        digits=4
    )
)


rf_auc = roc_auc_score(y_test, y_rf_prob)

print(f"\nROC AUC Score : {rf_auc:.4f}")
```

```python
# ==========================================
# 6. LIGHTGBM MODEL
# ==========================================

lgbm_model = LGBMClassifier(
    class_weight='balanced',
    random_state=9001,
    verbose=-1,
    n_estimators=100,
    max_depth=2,
    learning_rate=0.1
)

lgbm_model.fit(X_train, y_train)

y_lgbm_pred = lgbm_model.predict(X_test)

# Probability predictions
y_lgbm_prob = lgbm_model.predict_proba(X_test)[:,1]
```

```python
# ==========================================
# 7. LIGHTGBM EVALUATION
# ==========================================

lgbm_cm = confusion_matrix(
    y_test,
    y_lgbm_pred
)

print("LightGBM Confusion Matrix")
print(lgbm_cm)


disp = ConfusionMatrixDisplay(
    confusion_matrix=lgbm_cm,
    display_labels=bc_data.target_names
)

disp.plot(cmap=plt.cm.Blues)

plt.title("LightGBM Confusion Matrix")

plt.show()


print("\nClassification Report\n")

print(
    classification_report(
        y_test,
        y_lgbm_pred,
        digits=4
    )
)


lgbm_auc = roc_auc_score(
    y_test,
    y_lgbm_prob
)

print(f"\nROC AUC Score : {lgbm_auc:.4f}")
```
